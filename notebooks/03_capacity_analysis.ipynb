```python
beds= pd.read_excel(
    dt_path, sheet_name="Open Overnight",
    header=12
)
beds.columns=["Unused_0", "Year", "Period", "Org_Name", "Available_Total", "Available_General_Acute", "Available_Learning_Disabilities", "Available_Maternity", "Available_Mental_Illness", "Unused_9", "Occupied_Total",
              "Occupied_General_Acute", "Occupied_Learning_Disabilities", "Occupied_Maternity", "Occupied_Mental_Illness", "Unused_15", "Occupancy_Total", "Occupancy_General_Acute", "Occupancy_Learning_Disabilities", "Occupancy_Maternity", "Occupancy_Mental_Illness", "Unused_21", "Unused_22"]
print(beds.shape)
print(beds.columns.tolist())
beds[["Year", "Period", "Org_Name", "Available_Total", "Occupied_Total", "Occupancy_Total"]].head()
```

```python
beds=beds.drop(
    columns=["Unused_0", "Unused_9", "Unused_15", "Unused_21", "Unused_22"])
beds=beds[beds["Year"]!="Year"].copy()
beds=beds.reset_index(drop=True)
print(beds.shape)
print(beds.isna().sum())
beds.head()
```
